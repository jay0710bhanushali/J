```python
import numpy as np
class fuzzylogic:

  def __init__(self, a, b, c, d):
    self.left = a
    self.centerleft = b
    self.centerright = c
    self.right = d

  def membership_trapezoid(self, x):
    if x <= self.centerleft and x >= self.left:
      return (x - self.left) / (self.centerleft - self.left)
    elif self.centerleft <= x <= self.centerright:
      return 1
    elif x <= self.right and x >= self.centerright:
      return (self.right - x) / (self.right - self. centerright)
    else:
      return 0

  def area_trapezoid(self):
    return ((self.centerleft - self.left) + (self.right - self.centerright)) * (self.centerright - self.centerleft) / 2

  def centroid(self):
    numerator = ((self.right - self.centerright) * (self.centerright + self.right)) + ((self.centerleft - self.left) * (self.left + self.centerleft))
    denominator = 2 * ((self.centerleft - self.left) + (self.right - self.centerright))
    return numerator / denominator if denominator != 0 else 0

#fuzzy sets
temp_below_avg = fuzzylogic(15, 20, 30, 40)
temp_low = fuzzylogic(-5, 10, 15, 25)
press_below_avg = fuzzylogic(1.25, 1.50, 2.5, 2.75)
press_low = fuzzylogic(0.25, 0.50, 1.5, 1.75)
heat_power_med_high = fuzzylogic(1.25, 1.50, 2.5, 2.75)
heat_power_high = fuzzylogic(4.25, 4.5, 5.5, 5.75)
valve_open_med_low = fuzzylogic(1.25, 1.50, 2.5, 2.75)
valve_open_small = fuzzylogic(0.25, 0.50, 1.5, 1.75)

#input values
temp_input = 17.5
press_input = 1.3

z1 = min(temp_below_avg.membership_trapezoid(temp_input), press_below_avg.membership_trapezoid(press_input))
z2 = min(temp_low.membership_trapezoid(temp_input), press_low.membership_trapezoid(press_input))

print("Z1 = ", z1)
print("Z2 = ", z2)
```

```python
#defuzzification
#heating power
num_c1 = ((z1 * heat_power_med_high.area_trapezoid() * heat_power_med_high.centroid()) + (z2 * heat_power_high.area_trapezoid() * heat_power_high.centroid()))
den_c1  = ((z1 * heat_power_med_high.area_trapezoid()) + (z2 * heat_power_high.area_trapezoid()))

#valve opening
num_c2 = ((z1 * valve_open_med_low.area_trapezoid() * valve_open_med_low.centroid()) + (z2 * valve_open_small.area_trapezoid() * valve_open_small.centroid()))
den_c2 = ((z1 * valve_open_med_low.area_trapezoid()) + (z2 * valve_open_small.area_trapezoid()))

#crisp outputs
c1 = num_c1 / den_c1 if den_c1 != 0 else 0
c2 = num_c2 / den_c2 if den_c2 != 0 else 0

print("C1 (heating power) = ", c1)
print("C2 (valve opening) = ", c2)
```
